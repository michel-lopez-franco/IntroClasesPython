# Notebook de ejemplo

Un notebook sencillo con una clase `Rectangulo`.

```python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura
```

## Crear objetos

```python
r = Rectangulo(4, 3)
print(f"Área: {r.area()}")
```

## Gráfica de áreas

```python
#| label: fig-areas
#| fig-cap: "Área de rectángulos con altura 3"
import matplotlib.pyplot as plt

bases = range(1, 11)
plt.plot(bases, [Rectangulo(b, 3).area() for b in bases], marker="o")
plt.xlabel("base")
plt.ylabel("área")
plt.show()
```
